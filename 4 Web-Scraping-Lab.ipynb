## Extract information from the given web site
You will extract the data from the below web site: <br> 


```python
#this url contains the data you need to scrape
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/datasets/Programming_Languages.html"
```

The data you need to scrape is the **name of the programming language** and **average annual salary**.<br> It is a good idea to open the url in your web broswer and study the contents of the web page before you start to scrape.


Import the required libraries


```python
#your code goes here
import pandas as pd
import requests  
from bs4 import BeautifulSoup  
```

Download the webpage at the url


```python
#your code goes here
# The 'url' variable is already defined in the cell above as:
# "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/datasets/Programming_Languages.html"

response = requests.get(url)
data = response.text
```

Create a soup object


```python
#your code goes here
soup = BeautifulSoup(data, "html.parser")
```

Scrape the `Language name` and `annual average salary`.


```python
#your code goes here
# Find the table in the webpage
table = soup.find('table')

# Initialize a list to hold our scraped data records
scraped_data = []

# Loop through rows to extract language name and salary
for row in table.find_all('tr'):
    cols = row.find_all('td')
    
    # Check if data cells exist to avoid errors on headers or blank rows
    if len(cols) >= 5:
        # Based on the webpage layout:
        # Column index 1 is Language, Column index 3 is Average Annual Salary
        language = cols[1].getText().strip()
        salary = cols[3].getText().strip()
        
        # Skip the header row text explicitly if it matches the headers
        if language != "Language":
            scraped_data.append([language, salary])
            print(f"Language: {language} | Salary: {salary}")
```

Save the scrapped data into a file named *popular-languages.csv*


```python
# your code goes here
import csv  # <-- Add this line right here!

# Write the extracted data array into a CSV file
with open('popular-languages.csv', mode='w', newline='', encoding='utf-8') as file:
    writer = csv.writer(file)
    
    # Write the column headers
    writer.writerow(['Language Name', 'Average Annual Salary'])
    
    # Write the records
    writer.writerows(scraped_data)

print("Data successfully saved to popular-languages.csv!")
```
